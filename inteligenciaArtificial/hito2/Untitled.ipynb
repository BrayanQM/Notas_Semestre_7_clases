```python
# ============================================================
# 1. CARGAR LIBRERIAS y configuracion de semilla aleatoria 
# ============================================================

import pandas as pd
import numpy as np
import tensorflow as tf

from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error

np.random.seed(42)
tf.random.set_seed(42)
```

```python
# ============================================================
# 2. CARGAR DATASET
# ============================================================

df = pd.read_excel("datos.xlsx")


print("\n==========================================")
print("DATASET")
print("==========================================")

print(df.head())

print("\nCantidad de registros:", len(df))

print("\nColumnas:")
print(df.columns.tolist())

print("\nTipo de dato:")
print(df.dtypes)

print("\nEstadisticas:")
print(df.describe())
```

```python
# ============================================================
# 3. SEPARAR FEATURES Y TARGET 
# ============================================================

X = df[["num1","num2","num3"]]

y = df["resultado"]

# ============================================================
# 4. DIVISIÓN TRAIN / TEST
# ============================================================
#test 10%, train 90%

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=80,
    random_state=20
)
```

```python
# ============================================================
# 4. CREAR LA RED NEURONAL
# ============================================================
model = tf.keras.Sequential([

    # CAPA DE ENTRADA
    tf.keras.layers.Input(shape=(3,)),

    # CAPA OCULTA 1
    tf.keras.layers.Dense(15,activation="relu"),

    # CAPA OCULTA 2
    tf.keras.layers.Dense(9,activation="relu"),

    # CAPA OCULTA 3
    tf.keras.layers.Dense(3,activation="relu"),

    # CAPA DE SALIDA
    # OJO: Sin función de activación, esto permite realizar REGRESIÓN.
    tf.keras.layers.Dense(1)
])

model.compile(

    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.01
    ),
    loss="mean_squared_error",
    metrics=["mean_absolute_error"]
)


# MOSTRAR ARQUITECTURA
model.summary()
```

```python
# ============================================================
# 5. ENTRENAR
# ============================================================

history = model.fit(

    X_train,
    y_train,

    epochs=150,

    batch_size=8,

    validation_split=0.10,

    verbose=1
)
```

```python
# ============================================================
# 6. EVALUAR EL MODELO
# ============================================================

loss, mae = model.evaluate(

    X_test,
    y_test,

    verbose=0
)

# ============================================================
# 7. REALIZAR PREDICCIONES CON EL DATASET TEST
# ============================================================

predicciones = model.predict(
    X_test,
    verbose=0
).flatten()

for i in range(len(X_test)):

    entrada = X_test.iloc[i].values

    real = y_test.iloc[i]

    prediccion = predicciones[i]

    prediccion_redondeada = round(
        prediccion
    )

    print(
        f"Entrada: "
        f"| Real={real:.0f} "
        f"| Predicción={prediccion:.4f} "
        f"| Redondeada={prediccion_redondeada}"
    )
```
